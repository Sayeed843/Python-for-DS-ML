# Seaborn Distribution Plots

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
sns.displot(data=tips, x='total_bill', kde=True)
plt.show()
```

```python
sns.displot(data=tips, x='total_bill', kde=True, bins=50)
plt.show()
```

```python
# jointplot
sns.jointplot(x='total_bill', y='tip', data=tips, height=8)
plt.show()
```

```python
# jointplot
sns.jointplot(x='total_bill', y='tip', data=tips, kind='hex', height=8)
plt.xlabel('Total Bill', fontsize=14)
plt.ylabel('Tips', fontsize=14)
plt.show()
```

```python
# jointplot
sns.jointplot(x='total_bill', y='tip', data=tips, kind='reg', height=8)
plt.xlabel('Total Bill', fontsize=14)
plt.ylabel('Tips', fontsize=14)
plt.show()
```

```python
# jointplot
sns.jointplot(x='total_bill', y='tip', data=tips, kind='kde', height=8)
plt.xlabel('Total Bill', fontsize=14)
plt.ylabel('Tips', fontsize=14)
plt.show()
```

```python
# jointplot
sns.jointplot(x='total_bill', y='tip', data=tips, kind='kde', height=8, color= 'skyblue', fill=True)
plt.xlabel('Total Bill', fontsize=14)
plt.ylabel('Tips', fontsize=14)
plt.show()
```

```python
sns.pairplot(data=tips)
plt.show()
```

```python
sns.pairplot(data=tips, hue='sex')
plt.show()
```

```python
sns.pairplot(data=tips, hue='sex', palette='coolwarm')
plt.show()
```

# Seaborn Matrix Plots

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset('tips')
flights = sns.load_dataset('flights')
```

```python
tips.head()
```

```python
flights.head()
```

```python
tips.corr(numeric_only=True)
```

```python
tc = tips.corr(numeric_only=True)
```

```python
fig = plt.figure(figsize=(12,6))
sns.heatmap(data=tc, annot=True)
plt.show()
```

```python
fig = plt.figure(figsize=(12,6))
sns.heatmap(data=tc, annot=True, cmap='coolwarm')
plt.show()
```

```python
flights
```

```python
flights.pivot_table(index='month', columns='year', values='passengers', observed=False)
```

```python
fp = flights.pivot_table(index='month', columns='year', values='passengers', observed=False)
```

```python
fp.head()
```

```python
fig = plt.figure(figsize=(12,6))
sns.heatmap(data=fp, annot=True, cmap='coolwarm')
plt.tight_layout()
plt.show()
```

```python
fig = plt.figure(figsize=(12,6))
sns.heatmap(data=fp, annot=True, cmap='coolwarm', linecolor='white', linewidths=1)
plt.tight_layout()
plt.show()
```

```python
sns.clustermap(data=fp)
plt.show()
```

```python
sns.clustermap(data=fp, standard_scale=1,cmap='coolwarm')
plt.show()
```

# Seaborn Grids

```python
iris = sns.load_dataset('iris')
```

```python
iris.head()
```

```python
iris['species'].unique()
```

```python
sns.pairplot(iris)
plt.show()
```

```python
g = sns.PairGrid(iris)
g.map(plt.scatter)
plt.show()
```

```python
g = sns.PairGrid(iris, height=1.5)
g.map_diag(sns.histplot, kde = True)
plt.show()
```

```python
g = sns.PairGrid(iris, height=2.0)
g.map_diag(sns.histplot, kde = True)
g.map_upper(plt.scatter)
plt.show()
```

```python
g = sns.PairGrid(iris, height=2.0)
g.map_diag(sns.histplot, kde = True)
g.map_upper(plt.scatter)
g.map_lower(sns.kdeplot, color='green', fill=True)
plt.show()
```

```python
g = sns.PairGrid(iris, height=2.0)
#g.map_diag(sns.histplot, kde = True)
#g.map_upper(plt.scatter)
g.map_lower(sns.kdeplot, color='blue', fill=True)
plt.show()
```
