# Python Crash Course - part-4

```python
def times2(var):
    return var*2
```

```python
times2(5)
```

```python
seq = [1,2,3,4,5]
```

```python
map(times2, seq)
```

```python
list(map(times2, seq))
```

```python
xyz = []
for x in seq:
    xyz.append(times2(x))
print(xyz)
```

```python
[num**2 for num in seq] # list coprehension
```
