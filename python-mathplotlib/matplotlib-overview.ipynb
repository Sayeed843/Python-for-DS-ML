# Matplotlib -part-1

```python
import matplotlib.pyplot as plt 
```

```python
import numpy as np
```

```python
x = np.linspace(0,5,11)
y = x ** 2
```

```python
x
```

```python
y
```

```python
# Functional Way
```

```python
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('My Test Graph')
plt.show()
```

```python
plt.subplot(1,2,1) #nrow, nclo, figure position
plt.plot(x, y, 'r')
plt.xlabel('Col-1')
plt.ylabel('Row-1')
plt.title('Plot Number-1')

plt.subplot(1,2, 2) #nrow, nclo, figure position
plt.plot(y,x, 'y')
plt.xlabel('Col-2')
plt.ylabel('Row-1')
plt.title('Plot Num-2')
plt.show()
```

```python
plt.subplot(1,2,2) #nrow, nclo, figure position
plt.plot(x, y, 'r')
plt.xlabel('Col-1')
plt.ylabel('Row-1')
plt.title('Plot Number-1')

plt.subplot(1,2, 1) #nrow, nclo, figure position
plt.plot(y,x, 'y')
plt.xlabel('Col-2')
plt.ylabel('Row-1')
plt.title('Plot Num-2')
plt.show()
```

```python
## Object oriented method
```

```python
fig = plt.figure()
axes = fig.add_axes([0.1, 0.1, 0.8, 0.8]) # l, B, W, H
axes.plot(x,y)
axes.set_xlabel('X-axis')
axes.set_ylabel('Y-Axis')
axes.set_title('Object oriented Graph')
plt.show()
```

```python
fig = plt.figure()
axes1 = fig.add_axes([0.1, 0.1, 0.8, 0.8])
axes2 = fig.add_axes([0.4, 0.3, 0.3, 0.3])
plt.show()
```

```python
fig = plt.figure()
axes1 = fig.add_axes([0.1, 0.1, 0.8, 0.8])
axes2 = fig.add_axes([0.6, 0.2, 0.25, 0.15])

axes1.plot(x,y)
axes1.set_title('Large Plot')
axes2.plot(y,x)
axes2.set_title('Smaller plot')
plt.show()
```

```python
# Part-2
```

```python
fig, axes = plt.subplots(figsize= (8,6), nrows=2, ncols=1)
axes[0].plot(x,y)
axes[1].plot(y,x)
plt.show()
```

```python
fig, axes = plt.subplots(figsize= (8,6), nrows=2, ncols=1)
axes[0].plot(x,y)
axes[0].set_xlabel('X value')
axes[0].set_ylabel('Y Value')
axes[0].set_title('First Plot')

axes[1].plot(y,x)
axes[1].set_xlabel('X value')
axes[1].set_ylabel('Y Value')
axes[1].set_title('Second Plot')
plt.tight_layout()
plt.show()
```

```python
### figsize, dpi 
```

```python
fig = plt.figure()

ax = fig.add_axes([0,0,1,1])
ax.plot(x, x**2, 'r', label= 'X Squared')
ax.plot(x, x**3,'y', label='X Cubed')
ax.set_title('Learning Legend')

ax.legend(loc=0)

#plt.tight_layout()
plt.show()
```
